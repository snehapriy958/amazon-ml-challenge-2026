```python
import pandas as pd

s1 = pd.read_csv("dataset/train/train_source1.tsv", sep="\t")
s2 = pd.read_csv("dataset/train/train_source2.tsv", sep="\t")
s3 = pd.read_csv("dataset/train/train_source3.tsv", sep="\t")
gt = pd.read_csv("dataset/train/train_ground_truth.tsv", sep="\t")

datasets = {
    "SOURCE 1": s1,
    "SOURCE 2": s2,
    "SOURCE 3": s3,
    "GROUND TRUTH": gt
}

for name, df in datasets.items():
    print("=" * 70)
    print(name)
    print("Shape:", df.shape)
    print("\nMissing values:")
    print(df.isna().sum())
    print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Ground Truth Match Analysis

gt = pd.read_csv(
    "dataset/train/train_ground_truth.tsv",
    sep="\t"
)

gt["num_matches"] = (
    gt["matched_entity_ids"]
    .fillna("")
    .astype(str)
    .apply(lambda x: 0 if x.strip() == "" else len(x.split(",")))
)

print("=" * 70)
print("GROUND TRUTH MATCH ANALYSIS")
print("=" * 70)

print("\nTotal Source 1 entities:", len(gt))

print("\nEntities with NO match:")
print((gt["num_matches"] == 0).sum())

print("\nEntities WITH at least one match:")
print((gt["num_matches"] > 0).sum())

print("\nMatch count distribution:")
print(gt["num_matches"].value_counts().sort_index())

print("\nBasic statistics:")
print(gt["num_matches"].describe())

print("\nMaximum number of matches for one Source 1 entity:")
print(gt["num_matches"].max())
```

```python
def count_source_matches(value):
    if pd.isna(value) or str(value).strip() == "":
        return 0, 0

    ids = str(value).split(",")

    s2_count = sum(x.strip().startswith("S2-") for x in ids)
    s3_count = sum(x.strip().startswith("S3-") for x in ids)

    return s2_count, s3_count


gt[["s2_matches", "s3_matches"]] = gt["matched_entity_ids"].apply(
    lambda x: pd.Series(count_source_matches(x))
)

print("=" * 70)
print("SOURCE 2 / SOURCE 3 MATCH ANALYSIS")
print("=" * 70)

print("\nSource 1 entities with S2 matches:")
print((gt["s2_matches"] > 0).sum())

print("\nSource 1 entities with S3 matches:")
print((gt["s3_matches"] > 0).sum())

print("\nSource 1 entities with BOTH S2 and S3 matches:")
print(
    ((gt["s2_matches"] > 0) & (gt["s3_matches"] > 0)).sum()
)

print("\nSource 1 entities with ONLY S2 matches:")
print(
    ((gt["s2_matches"] > 0) & (gt["s3_matches"] == 0)).sum()
)

print("\nSource 1 entities with ONLY S3 matches:")
print(
    ((gt["s2_matches"] == 0) & (gt["s3_matches"] > 0)).sum()
)

print("\nTotal S2 matches:", gt["s2_matches"].sum())
print("Total S3 matches:", gt["s3_matches"].sum())
```
